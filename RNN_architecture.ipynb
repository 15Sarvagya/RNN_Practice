```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Input
```

```python
model=Sequential()
model.add(Input(shape=(4,5)))
model.add(SimpleRNN(3))
model.add(Dense(1,activation="sigmoid"))
```

```python
model.summary()
```

```python
print(model.get_weights()[0].shape)
model.get_weights()[0]
```

This is my input layer, where I have 4 x 5 inputs and the next layer is a hidden layer so the weight between them is this.

```python
print(model.get_weights()[1].shape)
model.get_weights()[1]
```

This is my layer, where the hidden layer is attached to itself only such that it's sending feedback to itself only. That's why this is 3x3. 3 is the size of the hidden layer and it is attached to itself only and sending the feedback for the next time to all its other neurons. That's why it's 3x3. You will better understand in the notes.

```python
print(model.get_weights()[2].shape)
model.get_weights()[2]
```

These are the biases of the hidden layers.

```python
print(model.get_weights()[3].shape)
model.get_weights()[3]
```

These are the output layers which are converting the hidden to the output. That's why 3 x 1.
